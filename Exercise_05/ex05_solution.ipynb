## Task 1 — Two portfolios, before any optimizer

Slide 27 put an optimized portfolio next to a conventional Swiss balanced mandate.
Before touching an optimizer, evaluate two portfolios you already know how to handle.

The file gives volatilities and correlations, not a covariance matrix. Build one:

$$\Sigma = D\,R\,D, \qquad D = \operatorname{diag}(\sigma_1, \dots, \sigma_5),
\qquad \text{so} \quad \Sigma_{ij} = \rho_{ij}\,\sigma_i\,\sigma_j .$$

From there, Lecture 04's matrix notation applies unchanged:

$$E(R_P) = \mathbf{w}'\boldsymbol{\mu}, \qquad
\sigma_P = \sqrt{\mathbf{w}'\Sigma\,\mathbf{w}}, \qquad
SR = \frac{E(R_P) - R_f}{\sigma_P}.$$

Write `portfolio_stats(w)` returning those three numbers — every later task calls it —
and evaluate two fully invested portfolios. The first is the Pictet-60 style mandate of
slide 27: 30 % Developed World Equity, 10 % EM Equity, 5 % Commodities, 20 % Swiss
Equity, 35 % Swiss Government Bonds. The second is the equally weighted $1/N$ portfolio.

Then answer the question the two numbers pose: **which of the two would a mean-variance
investor rather hold, and does the answer depend on how risk-averse that investor is?**

*Deliverable: a 2×3 table of E(R), σ and SR, and one sentence on the ranking.*


```python
# SOLUTION
# STUB: build Sigma from the volatilities and the correlation matrix
Sigma = np.outer(sd, sd) * corr.values

# STUB: portfolio_stats(w) -> Series with E(R), sigma and the Sharpe ratio
def portfolio_stats(w):                                                       # KEEP
    er = w @ mu
    sdev = np.sqrt(w @ Sigma @ w)
    return pd.Series({"E(R)": er, "sigma": sdev, "SR": (er - rf) / sdev})
```

```python
# SOLUTION
W = {"Pictet-60 style": np.array([0.30, 0.10, 0.05, 0.20, 0.35]),             # KEEP
     "1/N":             np.repeat(1 / N, N)}                                  # KEEP

# STUB: one row of statistics per portfolio
comparison = pd.DataFrame({"Pictet-60 style": portfolio_stats(W["Pictet-60 style"]),
                           "1/N":             portfolio_stats(W["1/N"])}).T

print(comparison.to_string(float_format="{:.4f}".format))                     # KEEP
```

<!-- solution -->
The Pictet-60 style mandate returns 4.51 % at a volatility of 9.22 %, a Sharpe ratio of
**0.479**. The equally weighted portfolio returns 5.40 % at 10.69 %, a Sharpe ratio of
**0.496**. On these inputs, naïve equal weighting beats the balanced mandate — and it
does so before a single line of optimization has been run.

The second half of the question is the one that matters. $1/N$ is riskier in absolute
terms: 10.69 % against 9.22 %. That does **not** make it worse for a cautious investor,
and reading it that way is the mistake this task exists to catch. By the separation
property (slide 15), the Sharpe ratio is the entire criterion for ranking *risky*
portfolios; risk aversion enters afterwards, through $y^*$, when the investor mixes the
chosen risky portfolio with cash. A cautious investor who prefers 9.22 % of volatility
holds $1/N$ at $y = 0.86$ and ends up with the same volatility and a *higher* expected
return than the mandate delivers. The choice of risky portfolio is preference-free; only
the scaling is personal.

One further trap sits in the data file. The `Assumptions` sheet carries both a compound
and an arithmetic return for every class, and this exercise uses the arithmetic column
throughout — $\mathbf{w}'\boldsymbol{\mu}$ is a statement about one-period expected
returns, which is what the arithmetic mean estimates. Using the compound column pulls
every expected return down by roughly a variance term, and pulls it down most for the
most volatile classes, which quietly tilts the whole exercise toward bonds. This is the
same distinction as in Exercise 02: which mean belongs in a forecast.


### The optimizer you are given

The three functions below are complete. Read them before Task 2 — the arguments are the
economics, and you will be changing them rather than writing them.

`scipy.optimize.minimize(objective, w0, method="SLSQP", bounds=..., constraints=...)`
searches for the weight vector that makes `objective` as small as possible. Four pieces
matter:

- **`objective`** — a function of the weights returning one number. Minimizing
  $\mathbf{w}'\Sigma\mathbf{w}$ gives a minimum-variance portfolio; minimizing the
  *negative* Sharpe ratio maximizes it, since there is no `maximize`.
- **`w0`** — where the search starts. Equal weights, here.
- **`constraints`** — a list of dictionaries. `{"type": "eq", "fun": f}` demands
  $f(\mathbf{w}) = 0$. Being fully invested is $\sum_i w_i - 1 = 0$; hitting a target
  return is $\mathbf{w}'\boldsymbol{\mu} - \text{target} = 0$.
- **`bounds`** — a range per weight. This is where the short-sale ban lives: $(0, 1)$
  per asset forbids negative weights, and widening the range to $(-2, 3)$ permits them.
  Note that "long-only" is a *bounds* change, not an extra constraint.

`SLSQP` is the method that handles equality constraints and bounds together. It is a
local search, so it reports whether it converged; all three functions return `None`
rather than a wrong answer when it did not. A target return no portfolio can reach — say
12 % when the best single asset offers 8.65 % and shorting is banned — is exactly such a
case, and your code has to expect it.


```python
FULLY_INVESTED = {"type": "eq", "fun": lambda w: w.sum() - 1.0}
BOUNDS = {True: [(0.0, 1.0)] * N,        # long-only
          False: [(-2.0, 3.0)] * N}      # shorts and leverage permitted


def _solve(objective, long_only, extra=()):
    """Minimise `objective` over fully invested weights. None if SLSQP fails."""
    res = minimize(objective, np.repeat(1 / N, N), method="SLSQP",
                   bounds=BOUNDS[long_only],
                   constraints=[FULLY_INVESTED, *extra],
                   options={"ftol": 1e-14, "maxiter": 3000})
    return res.x if res.success else None


def min_variance_portfolio(long_only=True):
    """The MVP: lowest variance of all fully invested portfolios."""
    return _solve(lambda w: w @ Sigma @ w, long_only)


def min_variance_at(target, long_only=True):
    """Lowest variance among portfolios with expected return exactly `target`."""
    hit_target = {"type": "eq", "fun": lambda w: w @ mu - target}
    return _solve(lambda w: w @ Sigma @ w, long_only, (hit_target,))


def tangency_portfolio(mu_vec, long_only=True):
    """The maximum-Sharpe portfolio for the expected returns `mu_vec`."""
    return _solve(lambda w: -(w @ mu_vec - rf) / np.sqrt(w @ Sigma @ w), long_only)
```

## Task 2 — Trace the frontier, then find the tangency portfolio

Slide 8 described Markowitz as four steps. Steps 2 and 3 are now two function calls in a
loop.

Write `trace_frontier(long_only, top, n=120)`: for each of `n` target returns, minimize
variance subject to hitting that target, and collect the resulting volatility, Sharpe
ratio and gross exposure $\sum_i |w_i|$.

**Start the grid at the minimum-variance portfolio's expected return.** Below it, every
portfolio is dominated — there is another with the same volatility and a higher expected
return (slide 6) — and step 4 of the procedure is precisely to discard that branch.
Tracing from zero instead is the commonest way to end up presenting a "frontier" whose
lower half no investor would ever hold.

Trace twice: **long-only** up to the highest single-asset expected return, and
**unconstrained** up to 12 %. Then locate the tangency portfolio under each bounds
regime with `tangency_portfolio`, print the two side by side with their gross exposures,
and read the gross exposure along the unconstrained frontier at target returns of 4, 6,
8, 10 and 12 %.

Two questions to answer from the output. **What does the short-sale ban cost at the
tangency point?** And **what is the unconstrained frontier buying with its extra
exposure** as the target return rises?

*Deliverable: the figure, the two tangency portfolios side by side, and the
gross-exposure table.*


```python
# SOLUTION
def trace_frontier(long_only, top, n=120):                                    # KEEP
    # STUB: begin the grid at the MVP's expected return, not at zero
    w_mvp = min_variance_portfolio(long_only)
    targets = np.linspace(w_mvp @ mu, top, n)

    # STUB: minimise variance at each target; skip any the solver cannot reach
    rows = []
    for target in targets:
        w = min_variance_at(target, long_only)
        if w is None:
            continue                     # this target is out of reach
        sigma = np.sqrt(w @ Sigma @ w)   # the solve pins E(R) to target
        rows.append({"target": target,
                     "sigma": sigma,
                     "SR": (target - rf) / sigma,
                     "gross": np.abs(w).sum()})
    return pd.DataFrame(rows)                                                 # KEEP


lo = trace_frontier(long_only=True,  top=mu.max())                            # KEEP
un = trace_frontier(long_only=False, top=0.12)                                # KEEP
print(f"long-only frontier: {len(lo)} points, top target {lo['target'].max():.2%}")
print(f"unconstrained     : {len(un)} points, top target {un['target'].max():.2%}")
```

```python
# SOLUTION
# STUB: the tangency portfolio under each bounds regime
w_tp = {"long-only": tangency_portfolio(mu, long_only=True),
        "unconstrained": tangency_portfolio(mu, long_only=False)}

# STUB: the two weight vectors as one table, and their statistics as another
tp_weights = 100 * pd.DataFrame(w_tp, index=RISKY).T
tp_stats = pd.DataFrame({"long-only":     portfolio_stats(w_tp["long-only"]),
                         "unconstrained": portfolio_stats(w_tp["unconstrained"])}).T
tp_stats["gross"] = tp_weights.abs().sum(axis=1) / 100

print("weights, %")                                                           # KEEP
print(tp_weights.round(1).to_string(), "\n")                                 # KEEP
print(tp_stats.to_string(float_format="{:.4f}".format))                       # KEEP

# STUB: gross exposure along the unconstrained frontier at five target returns
wanted = [0.04, 0.06, 0.08, 0.10, 0.12]
nearest = pd.Index(un["target"]).get_indexer(wanted, method="nearest")
ladder = un.iloc[nearest].copy()                 # the traced points nearest those
ladder[["target", "sigma", "gross"]] *= 100      # read the three as percentages

print("\ngross exposure, unconstrained frontier")
print(ladder.round({"target": 2, "sigma": 2, "SR": 3, "gross": 0})
             .to_string(index=False))
```

```python
fig1, ax = plt.subplots(figsize=(7.4, 5.2))

# Long-only underneath, unconstrained dashed on top: where the two coincide the
# dashed line sits on the thick one, and the divergence above is unmistakable.
ax.plot(100 * lo["sigma"], 100 * lo["target"], color=FHNW["navy"], lw=4.0,
        solid_capstyle="round", label="Long-only frontier")
ax.plot(100 * un["sigma"], 100 * un["target"], color=FHNW["blue"], ls="--", lw=1.6,
        label="Unconstrained frontier")

s_tp = portfolio_stats(w_tp["long-only"])
cal_x = np.linspace(0, 100 * lo["sigma"].max() * 1.3, 50)
ax.plot(cal_x, 100 * rf + s_tp["SR"] * cal_x, color=FHNW["orange"], lw=1.4,
        label="CAL through the tangency portfolio")

ax.scatter(100 * sd, 100 * mu, s=34, color="#7A7A7A", zorder=3)
x_max = 100 * lo["sigma"].max() * 1.35
# Five fixed points, so the labels are placed by hand rather than by a rule that
# would put one of them on top of the frontier. "EM Equity" is the slide-27 name.
PLACE = {"Developed World Equity":  ("Developed World Equity", (9, -3), "left"),
         "Emerging Markets Equity": ("EM Equity", (0, -9), "center"),
         "Commodities":             ("Commodities", (0, -9), "center"),
         "Swiss Equity":            ("Swiss Equity", (0, -9), "center"),
         "Swiss Government Bonds":  ("Swiss Government Bonds", (9, -3), "left")}
for k, name in enumerate(RISKY):
    label, offset, align = PLACE[name]
    ax.annotate(label, (100 * sd[k], 100 * mu[k]), fontsize=8, color="#555555",
                ha=align, va="center" if align == "left" else "top",
                xytext=offset, textcoords="offset points")

w_mvp = min_variance_portfolio(long_only=True)
s_mvp = portfolio_stats(w_mvp)
ax.scatter([100 * s_mvp["sigma"]], [100 * s_mvp["E(R)"]], s=70, marker="s",
           color=FHNW["green"], zorder=4, label="MVP, long-only")
ax.scatter([100 * s_tp["sigma"]], [100 * s_tp["E(R)"]], s=130, marker="*",
           color=FHNW["red"], zorder=4, label="Tangency portfolio")
ax.scatter([0], [100 * rf], s=34, color=FHNW["orange"], zorder=4)

ax.set_xlabel("Volatility, % p.a.")
ax.set_ylabel("Expected return, % p.a.")
ax.set_xlim(0, x_max)
ax.set_ylim(0, 13)
ax.legend(loc="lower right", fontsize=9)
```

31.7 % EM Equity, 11.7 % Commodities, 1.0 % Swiss Equity, 32.8 % Swiss Government
Bonds**, expected return 5.23 %, volatility 10.11 %, Sharpe ratio **0.507**, gross
exposure exactly 100 % in both cases. Dropping the short-sale ban changes nothing,
because at the tangency point the ban was not binding: no weight wanted to be negative
in the first place.

That is the first answer, and it is worth sitting with, because it contradicts the
intuition most people bring to slide 9. A constraint can only hurt — but it can only
hurt *where it binds*. Here the optimum is a genuinely diversified portfolio, roughly a
third in bonds and two thirds spread across three equity and commodity blocks, and the
long-only frontier and the unconstrained frontier touch at that point.

The second answer is in the gross-exposure column, and it is where the pathology
actually lives:

| Target E(R) | 3.96 % | 6.04 % | 8.03 % | 10.01 % | 12.00 % |
|---|---|---|---|---|---|
| σ | 7.68 % | 11.74 % | 15.78 % | 19.88 % | 24.01 % |
| Sharpe | 0.503 | 0.506 | 0.503 | 0.499 | 0.496 |
| Gross exposure | 100 % | 102 % | 113 % | 170 % | **227 %** |

(The five rows are the traced grid points nearest those targets, so the returns are not
exactly round.) Push the target from 4 % to 12 % and gross exposure more than doubles: at
the top the portfolio is 75 % Developed World Equity and 76 % EM Equity, financed by
shorting 49 % of Swiss government bonds and 14 % of Swiss equity. Meanwhile the Sharpe
ratio peaks at **0.507** at the tangency point itself, 5.23 %, and falls away on either
side — to 0.496 by the time the target reaches 12 %. The extra 127 percentage points of
gross exposure buy expected return, and they buy it at a slightly worse rate than simply
levering the tangency portfolio along the CAL would. That is the honest version of the
optimizer's output is not obviously wrong at the tangency, but every step you take away
from it in pursuit of return is paid for in positions that a mandate would never permit.

Two mistakes are worth having made once. The first is starting the target grid at zero
instead of at the MVP: the curve then doubles back on itself and the lower arm, which no
investor would hold, gets plotted as though it were part of the answer. The second is
forgetting that `min_variance_at` returns `None` — a long-only portfolio cannot reach
9 % when the best single asset offers 8.65 %, and code that assumes success crashes
somewhere in the last fifth of the grid.

There is also a trap you avoided only because the data file was adjusted, and it is a
real one. The published LTCMA figure for Swiss cash is 1.20 %, which is *above* the
unconstrained MVP's expected return of 0.75 %. When $R_f$ sits above the MVP, no ray
from $R_f$ ever touches the efficient branch: the Sharpe ratio climbs monotonically
along it toward an asymptote it never reaches, and there is no tangency portfolio at
all. The closed-form solution $\mathbf{w} \propto \Sigma^{-1}(\boldsymbol{\mu} -
R_f\mathbf{1})$ still returns *something* — the mirror portfolio on the dominated lower
branch, +165 % in government bonds against −61 % in developed equity, with a Sharpe
ratio of −0.49. It looks like an answer. It is a symptom.


## Task 3 (optional) — Move one input by less than its standard error

*For groups that finish Tasks 1 and 2 with time to spare. Nothing later depends on it,
and we work through the result together in the walkthrough either way.*

Slide 25 put a number on how badly expected returns are known. Estimated from $T$ years
of data, the standard error of a mean return is

$$SE = \frac{\sigma}{\sqrt{T}},$$

so twenty-five years of Developed World Equity at $\sigma = 16.14\ \%$ pins its
expected return down to about **3.2 percentage points** either way. Now perturb the
inputs by a fraction of that.

Take each of the five asset classes in turn, move its expected return by
$\pm 0.5$ percentage points — roughly one sixth of that standard error — and re-solve
the **long-only** tangency portfolio. Ten re-solves against the base case, one table.

Report the resulting weight table, and for each bump the largest single weight change
against the base case. Put the five standard errors beside it, so the comparison is
between how much the input was allowed to move and how much the output actually did.

Then answer: **is the sensitivity the same for every asset, and if not, what explains
the differences?**

*Deliverable: the weight table, the largest shift per bump, and the standard errors.*


```python
# SOLUTION
BUMP = 0.005                                                                  # KEEP

# STUB: re-solve the long-only tangency with one expected return moved by +-BUMP
def bumped_tangency(k, delta):       # asset k's E(R) moved, the rest left alone
    mu_bumped = mu.copy()
    mu_bumped[k] += delta
    return tangency_portfolio(mu_bumped, long_only=True)

weights = {"base case": tangency_portfolio(mu, long_only=True)}
for k, name in enumerate(RISKY):
    weights[f"{name} +0.5pp"] = bumped_tangency(k, +BUMP)
    weights[f"{name} -0.5pp"] = bumped_tangency(k, -BUMP)

bumps = pd.DataFrame(weights, index=RISKY).T                                  # KEEP

# STUB: the largest single weight change against the base case, per bump
shift = (bumps - bumps.loc["base case"]).abs().max(axis=1).drop("base case")

print((100 * bumps).round(1).to_string())                                     # KEEP
```

```python
# SOLUTION
# STUB: the standard error of a 25-year mean estimate, in percentage points
se_25y = pd.Series(sd / np.sqrt(25), index=RISKY)

print("largest weight shift, percentage points")                              # KEEP
print((100 * shift).round(1).sort_values(ascending=False).to_string())        # KEEP
print("\nstandard error of a 25-year mean estimate, percentage points")      # KEEP
print((100 * se_25y).round(2).to_string())                                    # KEEP

fig2, ax = plt.subplots(figsize=(7.4, 4.4))                                   # KEEP
order = shift.sort_values()                                                   # KEEP
colours = [FHNW["navy"] if s.endswith("+0.5pp") else FHNW["blue"]             # KEEP
           for s in order.index]                                              # KEEP
ax.barh(range(len(order)), 100 * order.values, color=colours)                 # KEEP
ax.set_yticks(range(len(order)))                                              # KEEP
ax.set_yticklabels(order.index, fontsize=8)                                   # KEEP
ax.set_xlabel("Largest weight change against the base case, percentage points")  # KEEP
handles = [plt.Rectangle((0, 0), 1, 1, color=FHNW["navy"]),                   # KEEP
           plt.Rectangle((0, 0), 1, 1, color=FHNW["blue"])]                   # KEEP
ax.legend(handles, ["E(R) raised by 0.5pp", "E(R) lowered by 0.5pp"],         # KEEP
          loc="lower right", fontsize=9)                                      # KEEP
```

The base case is the tangency portfolio from Task 2: 22.7 / 31.7 / 11.7 / 1.0 / 32.8.
Ten perturbations of half a percentage point each produce this:

| Bump | Largest weight change |
|---|---|
| Swiss Government Bonds −0.5pp | **32.8 pp** — bonds go from 32.8 % to 0 % |
| Swiss Government Bonds +0.5pp | **25.5 pp** — bonds go from 32.8 % to 58.3 % |
| Developed World Equity −0.5pp | 20.2 pp — from 22.7 % to 2.5 % |
| Swiss Equity +0.5pp | 18.3 pp — from 1.0 % to 19.3 % |
| Developed World Equity +0.5pp | 11.6 pp |
| EM Equity ±0.5pp | 7.1–7.3 pp |
| Commodities ±0.5pp | 4.4–4.5 pp |
| Swiss Equity −0.5pp | 1.0 pp |

Against standard errors of 3.2, 3.6, 3.5, 2.7 and 0.8 percentage points. The input moved
by 0.5 pp; the largest output moved by 32.8. Swiss government bonds are the extreme
case: their expected return is known to within 0.84 pp, and moving it by well under that
takes the allocation anywhere from nothing to nearly sixty percent of the portfolio.
This is Michaud's error maximization on real published inputs — the optimizer does not
know its inputs are estimates, so it spends its effort exploiting differences that are
smaller than the noise in them.

The asymmetry in the last two rows is the more instructive finding, and it is the answer
to the second question. Swiss Equity sits at 1.0 % in the base case, effectively at the
long-only floor. Raise its expected return by 0.5 pp and it jumps to 19.3 %; lower it by
the same amount and nothing happens at all — the weight was already as close to zero as
the constraint permits, so the optimizer has nowhere to send it. Sensitivity is not a
property of an asset; it is a property of where that asset currently sits relative to a
binding constraint. This is slide 9's two-sided effect made concrete: the short-sale ban
absorbs half of the instability and leaves the other half untouched, which is why
constrained portfolios *look* better behaved without the underlying estimation problem
having improved in the slightest.

The mistake to avoid is comparing the wrong pair of numbers. The point is not that a
0.5 pp bump changes the weights — of course it does. The point is the *ratio*: a
perturbation one sixth the size of the estimation error produces weight changes of
twenty to thirty percentage points. Scale the bump up to a full standard error, which is
what the data actually permits, and the base-case portfolio carries essentially no
information about which assets to hold. Slide 29's rough decision rule follows directly:
where no credible view on returns exists, methods that never estimate one — minimum
variance, risk parity, $1/N$ — are not a retreat from optimization but a reasonable
response to it.
